```python
Dict = {key1: value1, key2: value2}
1. It is a collection of key value pair.
2. We can not access the elements by using index.
3. Duplicate keys are not allowed, but duplicate value can be allowed.
4. Dictionary elements are comma seperated.
5. It is enclosed with curly bracket.
6. Dictionary is mutable.

keys >> immutable datatypes >> int, float, string, tuple
values >> Any datatype
```

# Functions

There are two types of functions
1. inbuilt functions e.g. min, max, sum, del, len
2. user defined functions

Functions:
    Functions is a block of code that can call or use wherever we wish to use.
    Functions are organized block of codes which are reusable.
    It will execute only when we do call it.
    

Declaration of Function: we create a function

Calling of a function: using a function.

```python
def function_name(parameters/Arguments): # defining a function

    block of code    # block of code

    return           # optional



function_name(parameters/Arguments)      # Calling the function to make a use of it.
```

```python
def func():
    print('We are learning user defined functions')
    return 'Hello Everyone'
```

```python
func()
```

```python
num = int(input('Enter a number: '))
fact = 1
if num < 0:
  print('Factorial of negative number is not defined.')
elif num == 0:
  print('Factorial of 0 is 1.')
else:
  for i in range(1,num+1):
    fact= fact*i
  print(f'Factorial of {num} is {fact}')
```

```python
def factorial(num):
    fact = 1
    if num < 0:
        print('Factorial of negative number is not defined.')
    elif num == 0:
        print('Factorial of 0 is 1.')
    else:
        for i in range(1,num+1):
            fact= fact*i
        print(f'Factor of {num} is {fact}')
```

```python
factorial(5)
```

```python
factorial(10)
```

```python
factorial(-5)
```

```python
factorial(0)
```

### 2. Default Arguments

These are the arguments which are declared in the paranthesis of a function itself.
It means if in case a user will failed to pass the parameter value while calling the function, it takes the by default value provided or declared at the time of defining a function.

```python
def add1(a,b=10):
    print(f'a == {a} and b == {b}' )
    add = a+b
    print(f'Addition of {a} and {b} is {add}')
```

```python
add1()
```

```python
add1(5)
```

```python
add1(5, 9)
```

```python
def addition(a,b,c=25): # a,b are positional arguments, c is default argument
    print(f'a ={a}, b={b}, c={c}')
    add = a+b+c
    return add
```

```python
addition(10,20,12)
```

### 3. Arbitrary Positional Argument

Variable length argument

*args >> *argument

Default data type of Args is tuple

This can be used when we are not sure about the exact no. of arguments required in our function.

```python
def summation(a,b,c,d,e,f,g):
    print('Our values are:', a,b,c,d,e,f,g)
```

```python
summation(1,2,3,4,5,6,7)
```

```python
summation(1,2,3,4,5,6)
```

```python
def sum1(*arg):
    print('Argument is:', arg)
```

```python
sum1(1)
```

```python
sum1(1,5,7,8,10,3,5,6,4)
```

```python
def addition(*arg):
    print('Argument is:', arg)
    add = 0
    for i in arg:
        add = add +i
    return add
```

```python
addition(1,5,7,8,10,3,5,6,4)
```

```python
addition(2,4)
```

```python
def addition(*sma):
    print('Argument is:', sma)
    add = 0
    for i in sma:
        add = add +i
    return add
```
